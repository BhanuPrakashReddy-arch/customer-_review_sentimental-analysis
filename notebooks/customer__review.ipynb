```python

import re

def clean_text(text):
    # Handle missing text safely
    if not isinstance(text, str):
        return ""

    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Replace repeated whitespace with a single space
    text = re.sub(r"\s+", " ", text).strip()

    return text

# Keep the original sentence column unchanged
df["cleaned_sentence"] = df["sentence"].apply(clean_text)

# Compare original and cleaned text
print(df[["sentence", "cleaned_sentence"]].head(10))
```

```python
sample_text = df["cleaned_sentence"].iloc[0]
print("original data:",sample_text)
tokens = sample_text.split()
print("Tokens:",tokens)
print("\n no of tokens:",len(tokens))
```

```python

# Normalize the cleaned sentences
def normalize_text(text):
    # Convert text to lowercase
    text = text.lower()

    # Remove extra whitespace
    text = " ".join(text.split())

    return text

# Create a new column without changing the original text
df["normalized_sentence"] = df["cleaned_sentence"].apply(normalize_text)

# Compare all three versions
print(
    df[["sentence", "cleaned_sentence", "normalized_sentence"]].head(10)
)
```

```python
sample = "  <b>AMAZING Product!</b>   Visit https://example.com  "

cleaned = clean_text(sample)
normalized = normalize_text(cleaned)

print("Original:   ", sample)
print("Cleaned:    ", cleaned)
print("Normalized: ", normalized)
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer
vectorizer = TfidfVectorizer(max_features=5000)

# Convert normalized text into numerical features
X = vectorizer.fit_transform(df["normalized_sentence"])

# Display the result
print("Feature matrix shape:", X.shape)

# Display some vocabulary words
print("\nFirst 20 vocabulary words:")
print(vectorizer.get_feature_names_out()[:20])

# Display the numerical features for the first sentence
print("\nFirst sentence vector:")
print(X[0].toarray())
```

```python

from sklearn.model_selection import train_test_split

X_train_text, X_test_text, y_train, y_test = train_test_split(
    df["normalized_sentence"],
    df["label"],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)

print("Training samples:", len(X_train_text))
print("Testing samples:", len(X_test_text))

print("\nTraining label distribution:")
print(y_train.value_counts())

print("\nTesting label distribution:")
print(y_test.value_counts())
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=5000)

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print("Training feature matrix:", X_train.shape)
print("Testing feature matrix:", X_test.shape)

print("\nFirst 20 vocabulary words:")
print(vectorizer.get_feature_names_out()[:20])
```

```python

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

print("Model training completed!")
```

```python

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

model_balanced = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

model_balanced.fit(X_train, y_train)

y_pred_balanced = model_balanced.predict(X_test)

print("Balanced model accuracy:",
      round(accuracy_score(y_test, y_pred_balanced) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_balanced))
```

```python

reviews = [
    "This product is amazing and I love it",
    "The product is okay, nothing special",
    "This is the worst product I have ever bought"
]

cleaned_reviews = [normalize_text(clean_text(review)) for review in reviews]
review_features = vectorizer.transform(cleaned_reviews)

predictions = model_balanced.predict(review_features)

for review, sentiment in zip(reviews, predictions):
    print("Review:", review)
    print("Predicted sentiment:", sentiment)
    print()
```

```python

import joblib
import os

os.makedirs("sentiment_model", exist_ok=True)

joblib.dump(
    model_balanced,
    "sentiment_model/logistic_regression.pkl"
)

joblib.dump(
    vectorizer,
    "sentiment_model/tfidf_vectorizer.pkl"
)

print("Model and vectorizer saved successfully!")

print(os.listdir("sentiment_model"))
```

```python

import joblib

loaded_model = joblib.load(
    "sentiment_model/logistic_regression.pkl"
)

loaded_vectorizer = joblib.load(
    "sentiment_model/tfidf_vectorizer.pkl"
)

new_review = "The product is excellent and I am very happy"

cleaned_review = normalize_text(clean_text(new_review))

review_features = loaded_vectorizer.transform([cleaned_review])

prediction = loaded_model.predict(review_features)[0]

print("Review:", new_review)
print("Predicted sentiment:", prediction)
```

```python

import os
import shutil

source_folder = "sentiment_model"
target_folder = "sentiment_analysis/models"

os.makedirs(target_folder, exist_ok=True)

for filename in [
    "logistic_regression.pkl",
    "tfidf_vectorizer.pkl"
]:
    source = os.path.join(source_folder, filename)
    target = os.path.join(target_folder, filename)

    if os.path.exists(source):
        shutil.copy2(source, target)
        print("Copied:", filename)
    else:
        print("File not found:", source)

print("\nModel folder contents:")
print(os.listdir(target_folder))
```
